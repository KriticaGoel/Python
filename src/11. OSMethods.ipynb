1. getcwd() -> get current working directories
2. mkdir() -> make directory
3. listdir() -> list all directory/file in current working directories
4. os.path.join() -> join multiple path together
5. os.path.exists(path) -> check path exists or not
6. os.path.isdir(file_path)
7. os.path.isfile(file_path)
8. os.path.abspath('1.Basic.ipynb') -> get absolute path

```python
import os
```

```python
print(os.getcwd())
```

```python
new_path= "new_folder"
os.mkdir(new_path)
print(f"Directory '{new_path}' created")
print(f"The current working directory is: {os.getcwd()}")
```

```python
print(os.listdir())
```

```python
file_name = "sample.txt"
folder_name = "new_folder"
os.makedirs(folder_name, exist_ok=True)
file_path = os.path.join(os.getcwd(),folder_name, file_name)
print(file_path)
```

```python
file_path=os.path.join(os.getcwd(),"nsa")
if(os.path.exists(file_path)):
    print(f"The path {file_path} exists.")
    if os.path.isfile(file_path):
        print(f"{file_path} is a file.")
    elif os.path.isdir(file_path):
        print(f"{file_path} is a directory.")
else:
    print(f"The path {file_path} does not exist.")
```
