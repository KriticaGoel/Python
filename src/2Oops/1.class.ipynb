```python
class Cartoon:
    def __init__(self,title,show):
        self.title = title
        self.show = show
    def display_info(self):
        return f"Cartoon Title: {self.title}, Show: {self.show}"
    
# if __name__ == "__main__":
#     my_cartoon = Cartoon("Tom and Jerry", "Tom and Jerry Show")
#     print(my_cartoon.display_info())

```

```python
mickey = Cartoon("Mickey Mouse","Mickey Mouse Clubhouse")
```

```python
print(mickey.display_info())
```

```python
spider = Cartoon("Spiderma","Spidy")
```

```python
print(spider.display_info())
```
