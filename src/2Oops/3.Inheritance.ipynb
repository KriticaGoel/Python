#### Types of Inheritance in python
1. Single inheritance
2. Multilevel Inheritance
3. Multiple Inheritance

#### Single inheritance -> parent -> Child

```python
class SuperHeross:

    def __init__(self, name, power):
        self.name = name
        self.power = power

    def show_hero(self):
        return f"{self.name} has the power of {self.power}"

class IronMan(SuperHeross):

    def __init__(self, name, power, suit_color):
        super().__init__(name, power)
        self.suit_color = suit_color

    def show_hero(self):
        base_info = super().show_hero()
        return f"{base_info} and wears a {self.suit_color} suit"
```

```python
hero1 = IronMan("Iron","Super Intelligence","Red and Gold")
```

```python
print(hero1.show_hero())
```

#### Multiple inheritance -> many parents -> one child

```python
class IronMans:

    def __init__(self, name, power):
        self.name = name
        self.power = power

    def iron_hero(self):
        return f"{self.name} has the power of {self.power}"
    
    def show_hero(self):
        return f"IronMans {self.name} has the power of {self.power}"  

class CaptainAmerica:

    def __init__(self, name, power):
        self.name = name
        self.power = power

    def show_hero(self):
        return f"CaptainAmerica {self.name} has the power of {self.power}"  

class Superhero(CaptainAmerica,IronMans):

    def __init__(self, name, power, suit_color):
        super().__init__(name, power)
        self.suit_color = suit_color

    def show_herossss(self):
        base_info = super().show_hero()
        return f"{base_info} and wears a {self.suit_color} suit"
```

```python
hero1= Superhero("Super","Super Intelligence","Red and Gold")
print(hero1.show_herossss())
print(hero1.iron_hero())
print(hero1.show_hero())
```
