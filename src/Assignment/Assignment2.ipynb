#### Task 1: Discount Rules

```python
order_amount = (input("Enter the order amount: "))

if(order_amount.isdigit()):
    order_amount = int(order_amount)
else:
    print("Invalid input. Please enter a numeric value.")
    exit()

tax=0.05
discount = 0.00

if order_amount >= 2000:
    discount = 0.15    
elif order_amount <2000 and order_amount >= 1500:
    discount = 0.10
elif order_amount <1500 and order_amount >= 1000:
    discount = 0.07
else:
    discount = 0.00

subtotal = order_amount - (order_amount * discount)
total = subtotal + (subtotal * tax)
print(f" Subtotal after discount: {subtotal} \n Tax is: {tax} \n Final amount is {total}")
```

#### Task 2: Process Multiple Orders

```python
orders = [1200,2500,800,1750,3000]
revenue=0.00
discount_count = 0
for order_amount in orders:
    tax=0.05
    discount = 0.00

    if order_amount >= 2000:
        discount = 0.15    
    elif order_amount <2000 and order_amount >= 1500:
        discount = 0.10
    elif order_amount <1500 and order_amount >= 1000:
        discount = 0.07
    else:
        discount = 0.00
    
    if(discount>0):
        discount_count+=1

    subtotal = order_amount - (order_amount * discount)
    total = subtotal + (subtotal * tax)
    print(f" Order amount : {order_amount} \n Discount%: {discount} \n Final Amount: {total} ")
    revenue += total
print(f" Total revenue from all orders is: {revenue}")
print(f" Total orders with discount applied: {discount_count}")
```

#### Task 3: User Menu 

```python
order_amount=[]
discount = 0.15

while True:
    print(f"press 1 to add order amount")
    print(f"press 2 to show all orders and total after discount")
    print(f"press q to exit")
    choice =input("Enter your choice:")
    if choice == '1':
        amount = input("Enter the order amount: ")
        if amount.isdigit():
            order_amount.append(int(amount))
        else:
            print("Invalid input. Please enter a numeric value.")
    elif choice == '2':
        total_revenue = 0.00
        for amount in order_amount:
            subtotal = amount - (amount * discount)
            print(f" Order amount : {amount} \n Discount%: {discount} \n Final Amount: {subtotal} ")
            total_revenue += subtotal
        print(f" Total revenue from all orders is: {total_revenue}")
    elif choice == 'q':
        print("Exiting the program.")
        break
    else:
        print("Invalid choice. Please try again.")

```

### Task 4: Loop Control with condition

```python
daily =[200,150,0,400,50,-1,300]
total_sales=0

for sale in daily:
    if sale<0:
        break
    elif sale == 0:
        continue
    else:
        total_sales += sale

print(f"Total sales for the day: {total_sales}")
```
