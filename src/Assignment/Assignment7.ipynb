#### Task1 Basic class and object creation 

```python
class Product:
    def __init__(self, name, price,category):
        self.name = name
        self.price = price
        self.category = category
    def getinfo(self):
        return f"{self.name} costs {self.price} and belongs to {self.category} category."
    
    def apply_discount(self, discount):
        self.price -= self.price * discount
    
apple = Product("Apple", 0.5, "Fruit")
print(apple.getinfo())  
banana = Product("Banana", 0.3, "Fruit")
print(banana.getinfo())


apple.apply_discount(0.1)
print(apple.getinfo())
```

### Task 2: Constructor and Encapsulation

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.__price = price

    def getinfo(self):
        return f"{self.name} costs {self.__price}."
    
    def getPrice(self):
        return self.__price
    
    def setPrice(self, price):
        if price < 0:
            raise ValueError("Price cannot be negative.")
        self.__price = price

apple = Product("Apple", 0.5)
apple.setPrice(0.4)
print(apple.getinfo())
```

#### Task 3: Inheritance

```python
class ElectronicProduct(Product):
    def __init__(self, name, price, warrenty_year):
        super().__init__(name, price)
        self.warrenty_year = warrenty_year

    def getinfo(self):
        return f"{self.name} costs {self.getPrice()} and has a warranty of {self.warrenty_year} years."
    

phone = ElectronicProduct("Smartphone", 699.99, 2)
print(phone.getinfo())
```

#### Task 4: Polymorphism

```python
class Laptop(Product):
    def getinfo(self):
        return f"The laptop {self.name} costs {self.getPrice()}."

class Mobile(Product):
    def getinfo(self):
        return f"The mobile {self.name} costs {self.getPrice()}."
    
laptops = [Laptop("Dell XPS 13", 999.99), Laptop("MacBook Pro", 1299.99)]
mobiles = [Mobile("iPhone 13", 799.99), Mobile("Samsung Galaxy S21", 699.99)]

for laptop in laptops:
    print(laptop.getinfo())
for mobile in mobiles:
    print(mobile.getinfo())
```

#### Task 5: Abstraction

```python
from abc import abstractmethod


class Payment:
    def __init__(self, amount):
        self.amount = amount

    @abstractmethod
    def process_payment(self):
        pass

class CreditCardPayment(Payment):
    def process_payment(self):
        return f"Processing credit card payment of ${self.amount}"  

class UpiPayment(Payment):
    def process_payment(self):
        return f"Processing UPI payment of ${self.amount}"

iciciCreditcard = CreditCardPayment(150)
print(iciciCreditcard.process_payment())
googlePay = UpiPayment(200)
print(googlePay.process_payment())
   
```

#### Task 6: Magic Method and Operator overloading

```python
class Product:
    def __init__(self, name, price,category):
        self.name = name
        self.price = price
        self.category = category
    
    def __str__(self):
        return f"Product(name={self.name}, price={self.price}, category={self.category})"
    
    def add(self, other):
        if isinstance(other, Product):
            return self.price + other.price
        return NotImplemented
    
apple = Product("Apple", 0.5, "Fruit")
banana = Product("Banana", 0.3, "Fruit")
print(apple.add(banana))
```

#### Task 7: Mini Project Simple Inventory system

```python
class Inventory:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def remove_product(self, name):
        for product in self.products:
            if product.name == name:
                self.products.remove(product)
                return True
        return False
    
    def get_total_value(self):
        return sum(product.price for product in self.products)

    def show_allProducts(self):
        return [str(product) for product in self.products]

class Store:
    def __init__(self, store_name):
        self.store_name = store_name
        self.inventory = Inventory()

    def add_new_product(self):
        print("Enter product details:")
        name = input("Name: ")
        price = float(input("Price: "))
        category = input("Category: ")
        new_product = Product(name, price, category)
        self.inventory.add_product(new_product)

    def show_summary(self):
        print(f"Total Items: {len(self.inventory.products)}")
        print(f"Total Value: ${self.inventory.get_total_value():.2f}")

store1 = Store("Tech Store")
store1.add_new_product()
store1.add_new_product()
store1.add_new_product()

store1.show_summary()

```
