```python
import numpy as np
```

```python
one_d_array = np.array(np.arange(1,10))
print(one_d_array)
two_d_array =np.array(np.arange(1,9)).reshape(2,-1)
print(two_d_array)
array_from_list= np.array([10,20,30,40,50])
print(array_from_list)
print("The shape of the array is:",one_d_array.shape)
print("The shape of the array is:",two_d_array.shape)
print("The shape of the array is:",array_from_list.shape)
print("The datatype of the array is:",one_d_array.dtype)
print("The datatype of the array is:",two_d_array.dtype)
print("The datatype of the array is:",array_from_list.dtype)
```

```python
A= np.array([10,20,30,40])
B=np.array([1,2,3,4])
print("The sum of the two arrays is:",A+B)
print("The difference of the two arrays is:",A-B)
print("The product of the two arrays is:",A*B)
print("The quotient of the two arrays is:",A/B)
print("The power of the two arrays is:",A**B)
print("The addition of the two arrays is:",np.add(A,B))
print("The subtraction of the two arrays is:",np.subtract(A,B))
print("The multiplication of the two arrays is:",np.multiply(A,B))
print("The division of the two arrays is:",np.divide(A,B))
print("The power of the two arrays is:",np.power(A,B))
```

```python
values=np.array([2,4,6,8,10])
print("the square root of each element in the array",np.sqrt(values))
print("The exponential of each element in the array",np.exp(values))
print("The natural logarithm of each element in the array",np.log(values))
print("The sum of all elements in the array is:",np.sum(values))
print(" The cumulative sum of the elements in the array is:",np.cumsum(values))
```

```python
data = np.array([[10,20,30],[40,50,60],[70,80,90]])
print(data)
print("The maximum value in the array is:",np.max(data))
print("The minimum value in the array is:",np.min(data))
print("The mean value of the array is:",np.mean(data))
print("Row sum of the array is:",np.sum(data,axis=1))
print("Column sum of the array is:",np.sum(data,axis=0))
```

```python
marks = np.array([78,85,90,66,72,88,95,60])
print("Mean of the marks is:",np.mean(marks))
print("Median of the marks is:",np.median(marks))
print("Standard deviation of the marks is:",np.std(marks))
print("Variance of the marks is:",np.var(marks))
print("Maximum marks is:",np.max(marks))
print("Minimum marks is:",np.min(marks))
print("Range minimum and maximum marks is:",np.ptp(marks))
```

```python
print("Sorting the marks in ascending order:",np.sort(marks))
print("Sorting the marks in descending order:",np.sort(marks)[::-1])
print("25% percentile of the marks is:",np.percentile(marks,25))
print("50% percentile of the marks is:",np.percentile(marks,50))
print("75% percentile of the marks is:",np.percentile(marks,75))
print("count above average marks is:",np.sum(marks>np.mean(marks)))
```

```python
sales = np.array([1200,1500,900,2000,1800,1700,1600])
print("The total weekly sales for the week is:",7* np.sum(sales))
print("The average daily sales for the week is:",np.mean(sales))
print("The maximum daily sales for the week is:",np.max(sales))
print("The minimum daily sales for the week is:",np.min(sales))
print("The standard deviation of daily sales for the week is:",np.std(sales))
print("Identify the days with sales above the average daily sales for the week:",np.where(sales>np.mean(sales)))
```
