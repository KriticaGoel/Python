3D array -> Tensor

Np.array() -> used to create multi dimension array

dtype parameter -> help us to decide the size of data wea re going to store in array
->np.int32 np.int64, np.float32, np.float64, np.bool

np.arange(1,10,2) -> used to create array using range function first parameter -> start from; second parameter -> end at; third parameter -> step

reshape(2,3) -> convert 1 d array to multi dimension array first parameter -> number of array; second parameter -> no. of element in each array
