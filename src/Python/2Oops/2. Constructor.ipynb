#### Instance Method
1. Instance method operates on object attribute
2. Instance method is going to use self keyword to access or modify the object attributes
3. Instance method called by using object.

```python
class SuperHeros:

    def __init__(self, name, power):
        self.name = name
        self.power = power

    def show_hero(self):
        return f"{self.name} has the power of {self.power}"
```

show_hero() - is instance method since value get change with each instance

```python
captonAmerica = SuperHeros("Captain America", "Super Strength")
ironMan = SuperHeros("Iron Man", "High Intelligence")
print(captonAmerica.show_hero())
print(ironMan.show_hero())
```

#### Class Method
1. Its is a method of whole class, which will remain same for all objects such kind of method is class method
2. work with class level attributes
3. Class method is going to use cls keyword instead of self
4. This method called using class name
5. To specify method as class method we need to use abbreviation as @classmethod

```python
class SuperHeross:

    origin ="SuperHeros"

    def __init__(self, name, power):
        self.name = name
        self.power = power

    def show_hero(self):
        return f"{self.name} has the power of {self.power}"
    
    @classmethod
    def show_origin(cls):
        return f"All heroes belong to {cls.origin}"
```

```python
print(SuperHeross.show_origin())
```

#### Static Method
1. Dont use cls nor self
2. work like normal function with belong to class only
3. Abbreviation @staticmethod
4. called by class name.

```python
class SuperHeross:

    origin ="SuperHeros"

    def __init__(self, name, power):
        self.name = name
        self.power = power

    def show_hero(self):
        return f"{self.name} has the power of {self.power}"
    
    @staticmethod
    def show_origin():
        return f"All heroes belong to {SuperHeross.origin}"
    
    @classmethod
    def show_origin(cls):
        return f"All heroes belong to {cls.origin}"
```

```python
print(SuperHeross.show_origin())
```
