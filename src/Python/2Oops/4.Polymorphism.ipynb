#### Types of poly morphism

1. Method Overriding -> Runtime Polymorphism
2. Method Overloading -> Compile time Polymorphism
3. Operator Overloading

```python
class Grandpa:
    def speak(self):
        return "Hello from Grand Pa"
    def eat(self):
        pass

class Parent(Grandpa):
    def speak(self):
        return "Hello from Parent"
    def eat(self):
        return "Parent eatting"
    
class Child(Grandpa):
    def speak(self):
        return "Hello from Child"
    def eat(self):
        return "Child eatting"
    
def attribute_provider(objects):
    return objects.speak(), objects.eat()
```

```python
grandpa= Grandpa()
print(attribute_provider(grandpa))
```

```python
attribute_provider(Parent())
```

```python
attribute_provider(Child())
```
